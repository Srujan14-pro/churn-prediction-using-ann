```python
from tensorflow.keras.models import load_model
import pickle

model = load_model('churn_model.h5')
with open('label_encoder.pkl', 'rb') as f:
    label_encoder = pickle.load(f)
with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
with open('one_hot_encoder.pkl', 'rb') as f:
    one_hot_encoder = pickle.load(f)
```

```python
input_data = {
    'Gender': 'Female',
    'Geography': 'France',
    'CreditScore': 600,
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
import pandas as pd

geo_encoded = one_hot_encoder.transform([['France']]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded, columns=one_hot_encoder.get_feature_names_out(['Geography']))
geo_encoded_df.head()
```

```python
# Use input_data directly since it's already a DataFrame
input_df = input_data
input_df
```

```python
input_df['Gender']=label_encoder.transform(input_df['Gender'])
input_df
```

```python
input_df = pd.concat([input_df.drop('Geography', axis=1), geo_encoded_df], axis=1)
input_df
```

```python
# Ensure input_df columns are in the same order as scaler was fitted
expected_columns = scaler.feature_names_in_
input_df_ordered = input_df[expected_columns]

input_scaled = scaler.transform(input_df_ordered)
input_scaled
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
pred_prob=prediction[0][0]
pred_prob

```

```python
if pred_prob > 0.5:
    print("The customer is likely to churn.")
else:
    print("The customer is unlikely to churn.")
    
```
